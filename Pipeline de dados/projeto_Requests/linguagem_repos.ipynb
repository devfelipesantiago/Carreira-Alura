# Conhecendo a requests

O método GET é um dos métodos HTTP mais comuns para fazer requisições em uma API. Ele é usado para obter dados de um recurso específico no servidor. Na biblioteca Requests em Python, o método GET é implementado por meio da função requests.get(), que aceita vários parâmetros para personalizar a requisição.

Aqui estão alguns dos parâmetros mais comuns do método GET da biblioteca Requests:

**url**: a URL do recurso que se deseja obter. Esse parâmetro é obrigatório e deve ser uma string contendo a URL que irá ser trabalhada.

**params**: um dicionário ou lista de tuplas contendo os parâmetros de consulta a serem incluídos na URL. Por exemplo, para fazer uma requisição GET para a API do GitHub e obter os repositórios de forma ordenada, pode-se passar o argumento params={”sort”: “name”} e ter os repositórios em ordem alfabética. Isso resultaria em uma URL como esta: “https://api.github.com/users/Username/repos?sort=name”

**headers**: um dicionário contendo os cabeçalhos HTTP a serem enviados com a requisição. Os cabeçalhos podem ser usados, por exemplo, para enviar informações de autenticação ou para definir o tipo de conteúdo esperado na resposta.

**auth**: uma tupla contendo as credenciais de autenticação a serem usadas na requisição. O primeiro elemento da tupla é o nome de usuário e o segundo é a senha. Essa opção é usada quando a API requer autenticação.

**timeout**: o tempo limite em segundos para aguardar a resposta da requisição. Se a resposta não for recebida dentro desse tempo limite, a biblioteca Requests gera um erro de tempo limite.

**verify**: um valor booleano que indica se deve ser verificado o certificado SSL da URL. Se definido como True, a biblioteca Requests verifica a validade do certificado. Se definido como False, a biblioteca não verifica o certificado.

O método GET é amplamente usado quando trabalhamos com APIs REST para obter informações de recursos específicos, e a biblioteca Requests oferece muitas opções para personalizar a requisição e manipular a resposta.

## Paginando os repositórios

```python
api_base_url = 'https://api.github.com'
owner = 'amzn'
url = f'{api_base_url}/users/{owner}/repos'
url
```

```python
repos_list = []

for page_num in range(1, 6):
    try:
        url_page = f'{url}?page={page_num}'
        response = requests.get(url_page, headers=headers)
        repos_list.append(response.json())
    except:
        repos_list.append(None)
```

```python
followers_list = []
page = 1

while True:

    url_page = f'{followers.url}?page={page}'
    response = requests.get(url_page, headers=headers)

    if len(response.json()) == 0:
        break

    followers_list.append(response.json())
    page += 1

followers_list
```
